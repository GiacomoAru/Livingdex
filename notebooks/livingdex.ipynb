```python
def search(name=None, shiny=None, captured=None, gigamax=None, box=None, row=10):
    def distance(small, long):
        if small in long and small != long:
            return 0.5
        return jellyfish.damerau_levenshtein_distance(small,long) + (1-jellyfish.jaro_winkler_similarity(small,long))


    dummy = livingdex.copy()

    if box != None:
        dummy['distance_box'] = dummy['box'].apply(lambda x: distance(box, x))
        dummy = dummy.sort_values('distance_box', kind='stable')
    if name != None:
        dummy['distance_id'] = dummy['id'].apply(lambda x: distance(name, x))
        dummy = dummy.sort_values('distance_id', kind='stable')


    if shiny != None:
        dummy = dummy.loc[dummy['shiny'] == shiny]
    if captured != None:
        dummy = dummy.loc[dummy['captured'] == captured]
    if gigamax != None:
        dummy = dummy.loc[dummy['gigamax'] == bool(gigamax)]
    
    return dummy.head(row)
```
