```python
import json
import pandas as pd
```

```python
livingdex = json.load(open('../data/pokemon/home---living-dex.livingdex.json', 'r'))['boxes']
```

```python
for box in livingdex:
    for pok in box['pokemon']:
        if pok['id'] == 'charizard':
            print(pok)
```

```python
new_liv = []

for box in livingdex:
    if not box['shiny']:
        new_liv.append(box)

    if 'shiny' in box:
        del box['shiny']

    for pok in box['pokemon']:
        if 'evs' in pok:
            del pok['evs']
        if 'ivs' in pok:
            del pok['ivs']
        if 'moves' in pok:
            del pok['moves']
```

```python
open('../data/pokemon/livingdex.json', 'w+').write(json.dumps(new_liv, indent=2))
```

```python
csv_dict = {'id':[], 'captured':[], 'shiny':[], 'box':[]}

for box in new_liv:
    for pk in box['pokemon']:
        
        csv_dict['captured'].append(pk['captured'])
        csv_dict['shiny'].append(pk['shiny'])
        id = pk['id']
        if len(pk['emblemMarks']) > 0:
            id += '-gigamax'
        csv_dict['id'].append(id)
        csv_dict['box'].append(box['title'].lower().replace(' ', '_'))


for i in range(len(csv_dict['id'])):
    if csv_dict['id'][i]+'-f' in csv_dict['id']:
        csv_dict['id'][i] = csv_dict['id'][i]+'-m'

livingdes_df = pd.DataFrame(csv_dict)
```
