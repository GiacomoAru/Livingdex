```python
import urllib.request

def scarica_dizionario_italiano(url="https://raw.githubusercontent.com/napolux/italian-words/master/italian.txt"):
    """
    Scarica un dizionario italiano da un URL usando solo la libreria standard.
    """
    try:
        with urllib.request.urlopen(url) as response:
            contenuto = response.read().decode("utf-8")
            dizionario = contenuto.splitlines()
            return [parola.lower() for parola in dizionario if parola.isalpha()]
    except Exception as e:
        raise Exception(f"Errore nel download del dizionario: {e}")

def filtra_parole(dizionario, n, lettere_richieste):
    lettere_richieste = set(lettere_richieste)
    return [
        parola for parola in dizionario
        if len(parola) == n and lettere_richieste.issubset(set(parola))
    ]

def parole_italiane_con_lettere(n, lettere_richieste):
    dizionario = scarica_dizionario_italiano()
    return filtra_parole(dizionario, n, lettere_richieste)
```

```python
av_names = [1,2,3,4,5,6,7,8,'8_1',9]
avs = []
for n in av_names:
    avs.append(pd.read_csv(f'./data/availability/{n}.csv').drop('Icon', axis=1))

remove_char = ['.', ' ', '♀', '♂', "'", '-']

for av in avs:
    new_names = []
    for i, v in av.iterrows():
        new_n = v['Name']
        
        if '(Galarian Form)' in new_n:
            new_n = new_n.replace('(Galarian Form)', '')
            new_n = new_n + '_galar'
        elif '(Alolan Form)' in new_n:
            new_n = new_n.replace('(Alolan Form)', '')
            new_n = new_n + '_alola'    
        elif '(Hisuian Form)' in new_n:
            new_n = new_n.replace('(Hisuian Form)', '')
            new_n = new_n + '_hisui' 
        elif '(Paldean Form)' in new_n:
            new_n = new_n.replace('(Paldean Form)', '')
            new_n = new_n + '_paldea'    
            
            
        elif '(Paldean Combat Form)' in new_n:
            new_n = new_n.replace('(Paldean Combat Form)', '')
            new_n = new_n + '_paldea'
        elif '(Paldean Blaze Form)' in new_n:
            new_n = new_n.replace('(Paldean Blaze Form)', '')
            new_n = new_n + '_paldea_fire'
        elif '(Paldean Aqua Form)' in new_n:
            new_n = new_n.replace('(Paldean Aqua Form)', '')
            new_n = new_n + '_paldea_water'
        
        new_n = new_n.lower()
        for c in remove_char:
            new_n = new_n.replace(c, '')
            
        new_names.append(new_n)
    
    av['Name'] = new_names


for i, av in enumerate(avs):
    av.to_csv(f'./data/availability/{i}_clean.csv', index=False)
```

```python
# Concatenazione mantenendo tutte le colonne
df_unico = pd.concat(avs, ignore_index=True, sort=False)
```
